```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os, sys

sys.path.append('../src')
df = pd.read_csv('../data/raw/Telco-Customer-Churn.csv')
print('Raw Data Shape:', df.shape)
df.head()
```

## 1. Data Cleaning & Quirks Handling
- Imputing TotalCharges whitespace strings for tenure=0
- Dropping non-predictive customerID

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].astype(str).str.strip(), errors='coerce')
df.loc[df['tenure'] == 0, 'TotalCharges'] = 0.0
df = df.drop(columns=['customerID'])
print('Missing Values:\n', df.isnull().sum())
```

## 2. Exploratory Data Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df['Churn'].value_counts(normalize=True).plot.bar(ax=axes[0], color=['#2b5c8f', '#d9534f'])
axes[0].set_title('Churn Class Imbalance')
sns.boxplot(data=df, x='Churn', y='MonthlyCharges', ax=axes[1])
axes[1].set_title('Monthly Charges vs Churn')
plt.tight_layout()
plt.show()
```
